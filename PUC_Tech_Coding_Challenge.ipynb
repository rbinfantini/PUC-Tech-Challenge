#### **Tarefa 1 - Conhecendo o Dataset**

#### Espaço para a resposta da Tarefa 1



```python
import pandas as pd

def carregar_dados():

    with open('games.csv', 'r', encoding='utf-8') as f:
        cabecalho_errado = f.readline().strip()
    cabecalho_corrigido = cabecalho_errado.replace('DiscountDLC count', 'Discount,DLC count')
    nomes_certos = cabecalho_corrigido.split(',')

    df = pd.read_csv('games.csv', skiprows=1, names=nomes_certos)
    df_limpo = df.drop_duplicates(subset=['AppID']).copy()

    # Filtro para isolar apenas os jogos
    filtro_software = 'Animation & Modeling|Design & Illustration|Game Development|Utilities|Software Training|Video Production|Audio Production|Photo Editing|Web Publishing'
    df_jogos = df_limpo[~df_limpo['Genres'].astype(str).str.contains(filtro_software, na=False, regex=True)].copy()

    return df_limpo, df_jogos


#a)

# Chamada da função reutilizável
df_limpo, df_jogos = carregar_dados()

total_bruto = df_limpo['AppID'].nunique()
total_jogos_real = df_jogos['AppID'].nunique()
total_apps = total_bruto - total_jogos_real

print("\na) Quantidade de Jogos")
print(f"   ► Total de aplicações : {total_bruto}")
print(f"   ► Total de jogos      : {total_jogos_real}")
print(f"   ► Total de apps       : {total_apps}")

#b)
total_windows = (df_limpo['Windows']).sum()
total_mac = (df_limpo['Mac']).sum()
total_linux = (df_limpo['Linux']).sum()
so_windows = df_jogos[(df_jogos['Windows'] == True) & (df_jogos['Mac'] == False) & (df_jogos['Linux'] == False)]
so_mac = df_jogos[(df_jogos['Windows'] == False) & (df_jogos['Mac'] == True) & (df_jogos['Linux'] == False)]
so_linux = df_jogos[(df_jogos['Windows'] == False) & (df_jogos['Mac'] == False) & (df_jogos['Linux'] == True)]

print("\nb) Compatibilidade por Sistema Operacional")
print(f"   ► Windows      : {total_windows} jogos")
print(f"   ► Mac          : {total_mac} jogos")
print(f"   ► Linux        : {total_linux} jogos")
print(f"   ► Só Windows   : {len(so_windows)} jogos")
print(f"   ► Só Mac       : {len(so_mac)} jogos")
print(f"   ► Só Linux     : {len(so_linux)} jogos")

#c)
media_geral = df_limpo['Price'].mean().round(2)
media_pagos = df_jogos[df_jogos['Price'] > 0]['Price'].mean().round(2)

print("\nc) Análise de Preço")
print(f"   ► Preço médio da plataforma  : $ {media_geral}")
print(f"   ► Preço médio de jogos pagos : $ {media_pagos}")
```

#### **Tarefa 3 - Primeiro Insight analítico**

A partir do conjunto de dados fornecido, identifique:

a) Proponha uma métrica de sucesso baseada em pelo menos duas variáveis (ex: número de avaliações positivas, tempo médio de jogo, recomendações, etc.)

b) Aplique sua métrica e liste os 10 jogos mais bem-sucedidos

c) Analise os resultados e responda:
Existe algum padrão em comum entre esses jogos? (ex: gênero, preço, popularidade, etc.)

#### Espaço para a resposta da Tarefa 3



```python
import pandas as pd
import matplotlib.pyplot as plt

# Chamada da função reutilizável
df_limpo, df_jogos = carregar_dados()

print('a) Métricas escolhidas')
print("""
As métricas escolhidas foram o total de reviews, a aprovação do jogo, a média global da Steam,
o preço dos jogos e as tags. Usarei a média que a plataforma do IMDb usa para decidir seu famoso
Top 250, a média Bayesiana.
Essa média também faz uso de um valor arbitrário chamado Peso de Confiança, que impede jogos
com pouquíssimas avaliações de passar por jogos com muitas avaliações, nesse caso, o valor
dela será 1800.
Também serão analisados separadamente o preço dos jogos e as tags deles. Tags são decididas
pelos jogadores e assim é possível perceber um padrão de preferência por mecânicas ou estilos.
""")

df_ranking = df_jogos.copy()

# Média Bayesiana
df_ranking['Total_Reviews'] = df_ranking['Positive'] + df_ranking['Negative']
df_ranking = df_ranking.sort_values(by='Total_Reviews', ascending=False).drop_duplicates(subset=['Name'])

media_global = df_ranking['Positive'].sum() / df_ranking['Total_Reviews'].sum()
m = 1800 # Peso de Confiança
df_ranking['Media_Bayesiana'] = (df_ranking['Positive'] + m * media_global) / (df_ranking['Total_Reviews'] + m)

# Seleção e formatação do Top 10
top_10 = df_ranking.nlargest(10, 'Media_Bayesiana').copy()
top_10['Name'] = top_10['Name'].str.replace('＊', '*') # Correção para o asterisco japonês

nomes_jogos = top_10['Name'].tolist()
top_10['Jogo'] = [f"{i} - {nome}" for i, nome in enumerate(nomes_jogos, 1)]
top_10['Nota_Str'] = (top_10['Media_Bayesiana'] * 100).apply(lambda x: f"{x:.2f}%") # Tranforma a nota em string % para a tabela
top_10['Principais_Tags'] = top_10['Tags'].astype(str).str.split(',').str[:5].str.join(', ')

print('\nb)')
print("--- TOP 10 JOGOS MAIS BEM AVALIADOS DA STEAM ---\n")
print(top_10[['Jogo', 'Price', 'Nota_Str', 'Principais_Tags']].to_string(
    index=False,
    justify='left',
    formatters={
        'Jogo': lambda x: f"{x:<40}",
        'Price': lambda x: f"{'Gratuito' if x == 0 else f'$ {x:.2f}':<10}",
        'Nota_Str': lambda x: f"{x:<10}",
        'Principais_Tags': lambda x: f"{str(x):<75}"
    }
))

print('\n')
print('c) Análise de Padrões')
print("""O único padrão facilmente detectável é o preço dos jogos já que 8 dos 10 custam menos de 10 dólares,
os jogos da lista também apresentam alta rejogabilidade, sendo jogos multiplayer ou com finais
diferentes. Apenas 3 jogos da lista possuem uma narrativa linear que o final não é alterado pela
criatividade do jogador, esses sendo Portal 1 e 2, e Half Life: Alix, porém, Portal é renomado pelos
seus puzzles e sua narrativa e Half Life: Alyx é um jogo vr que permite ao player um certo grau de
criatividade durante a campanha, mesmo que o final não seja alterado, o jogador chega nele com seus
próprios métodos. Portanto é justo afirmar que o top 10 é composto de jogos acessíveis que alimentam
a liberdade do player e tem alta rejogabilidade, com exceções de narrativa fechada que forçam a criatividade
do jogador.
""")

# Cortando nomes muito longos para caber no gráfico e pegando os valores reais
nomes_curtos = [nome[:25] + "..." if len(nome) > 25 else nome for nome in top_10['Name'].tolist()]
valores_nota = (top_10['Media_Bayesiana'] * 100).tolist()

plt.figure(figsize=(10, 6))

# Invertemos as listas [::-1] para o 1º lugar ficar no topo do gráfico
barras = plt.barh(nomes_curtos[::-1], valores_nota[::-1], color='#5C6BC0')

plt.xlabel('Nota Bayesiana (%)')
plt.title('Top 10 Jogos Mais Bem Avaliados da Steam\n(Média Bayesiana)', fontsize=14)

# Focando o eixo X no topo (já que as notas são altas)
plt.xlim(97, 100)

# Inserindo o valor percentual dentro da barra
for barra, valor in zip(barras, valores_nota[::-1]):
    plt.text(barra.get_width() - 0.1, barra.get_y() + barra.get_height()/2,
             f'{valor:.2f}%',
             ha='right', va='center', color='white', fontweight='bold')

plt.tight_layout()
plt.show()
```

# Desafio 3 - Criptografia

Você está atuando como analista de segurança em uma empresa que realiza auditorias internas para identificar vulnerabilidades em sistemas.

Durante uma análise, foi identificado que um sistema armazena CPFs parcialmente mascarados no formato:

`XXX.281.940-XX`

Porém, descobriu-se que o sistema de autenticação não possui proteção contra tentativas repetidas de acesso (como limite de tentativas ou CAPTCHA), o que pode permitir um ataque de força bruta.

**Recupere o CPF completo, seguindo a seguinte lista de tarefas:**

a) Implemente um algoritmo que gere combinações para os dígitos desconhecidos (`X`)

b) Para cada combinação gerada, utilize a função `autenticar(cpf)` para verificar se o CPF está correto  

c) Exiba:
- O CPF válido encontrado  
- A quantidade de tentativas necessárias até encontrá-lo  

```python
# Execute essa célula para utilizar a função de validação de CPF com um CPF real
# Curiosidade: Se quiser, pode substituir o CPF aqui pelo seu para testar seu algoritmo!

CPF_REAL = "919.281.940-01"

def autenticar(cpf):
    return cpf == CPF_REAL
```

## Espaço Para Resposta do Desafio 3:


```python
#a) XXX.281.940-XX

tentativas = 0
cpf_valido = None

# O loop externo testa os 3 primeiros digitos (000 a 999)
# O loop interno testa os 2 últimos digitos (00 a 99)
for primeiros_digitos in range(1000):
    for final in range(100):
        tentativas += 1
        cpf_teste = f"{primeiros_digitos:03d}.281.940-{final:02d}"
#b)
        if autenticar(cpf_teste):
            cpf_valido = cpf_teste
            break # Quebra o loop interno para economizar processamento

    if cpf_valido:
        break # Quebra o loop externo assim que achar o alvo
#c)
print(f'c) Validação de CPF')
print(f"O CPF válido: {cpf_valido}")
print(f"Quantidade de tentativas: {tentativas}")
```

```python
import requests
import json
import os

ARQUIVO_JSON = 'pokedex.json'

# FUNÇÕES AUXILIARES (HELPERS) E UX

def limpar_tela():
    try:
        from google.colab import output
        output.clear()
    except ImportError:
        # Fallback para rodar no terminal local (Windows/Linux/Mac)
        os.system('cls' if os.name == 'nt' else 'clear')

def buscar_pokemon(pokedex, alvo_digitado):
    e_numero = alvo_digitado.isdigit()
    for pokemon in pokedex:
        match_id = e_numero and pokemon['id'] == int(alvo_digitado)
        match_nome = not e_numero and pokemon['nome'].lower() == alvo_digitado.lower()
        if match_id or match_nome:
            return pokemon
    return None

def obter_nivel_valido(mensagem="Qual o nível dele? "):
    while True:
        try:
            nivel = int(input(mensagem))
            if 1 <= nivel <= 100:
                return nivel
            print("O Nível deve ser entre 1 e 100.")
        except ValueError:
            print("Digite um número válido.")

def carregar_pokedex():
    try:
        with open(ARQUIVO_JSON, 'r', encoding='utf-8') as f:
            return json.load(f)
    except FileNotFoundError:
        return []  # Primeira vez rodando, não há arquivo ainda
    except json.JSONDecodeError:
        print("Arquivo Pokédex corrompido. Iniciando nova Pokédex.")
        return []
    except Exception as e:
        print(f"Erro ao carregar Pokédex: {e}")
        return []

def salvar_pokedex(dados):
    try:
        with open(ARQUIVO_JSON, 'w', encoding='utf-8') as f:
            json.dump(dados, f, indent=4, ensure_ascii=False)
    except IOError as e:
        print(f"Erro ao salvar Pokédex: {e}")
    except Exception as e:
        print(f"Erro inesperado ao salvar: {e}")

def adicionar_pokemon(nome_digitado, nivel):
    pokedex = carregar_pokedex()

    url = f"https://pokeapi.co/api/v2/pokemon/{nome_digitado.strip().lower()}/"

    try:
        resposta = requests.get(url, timeout=10)

        if resposta.status_code != 200:
            print(f"O Pokemon '{nome_digitado}' não existe na base oficial.")
            return

        try:
            dados_api = resposta.json()
        except json.JSONDecodeError:
            print("Erro: Resposta inválida da API.")
            return

        id_oficial = dados_api['id']
        nome_oficial = dados_api['name'].capitalize()
        tipo_oficial = dados_api['types'][0]['type']['name'].capitalize()

        # Previne duplicatas verificando ID
        for pokemon in pokedex:
            if pokemon['id'] == id_oficial:
                print(f"Você já tem o {nome_oficial} no seu Pokedex.")
                return

        novo_pokemon = {
            "id": id_oficial,
            "nome": nome_oficial,
            "tipo": tipo_oficial,
            "nivel": nivel
        }

        pokedex.append(novo_pokemon)
        salvar_pokedex(pokedex)
        print(f"{nome_oficial} entrou no seu Pokedex.")

    except requests.exceptions.Timeout:
        print("Erro: A conexão demorou muito. Tente novamente.")
    except requests.exceptions.ConnectionError:
        print("Erro: Não foi possível conectar à PokeAPI. Verifique sua internet.")
    except requests.exceptions.RequestException as e:
        print(f"Erro de rede: {e}")

def listar_pokemons():
    pokedex = carregar_pokedex()

    if len(pokedex) == 0:
        print("\n  Seu Pokedex está vazio.")
        return

    print("\n┏━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━┓")
    print("┃                    POKEDEX                     ┃")
    print("┣━━━━━━┳━━━━━━━━━━━━━━┳━━━━━━━━━━━━━━┳━━━━━━━━━━━┫")

    for p in pokedex:
        id_formatado = str(p['id']).zfill(3)
        print(f"┃ #{id_formatado} ┃ {p['nome']:<12} ┃ {p['tipo']:<12} ┃ Nv: {p['nivel']:<4}  ┃")

    print("┗━━━━━━┻━━━━━━━━━━━━━━┻━━━━━━━━━━━━━━┻━━━━━━━━━━━┛\n")

def remover_pokemon(alvo_digitado):
    pokedex = carregar_pokedex()
    pokemon = buscar_pokemon(pokedex, alvo_digitado)

    if pokemon:
        confirmacao = input(f"\nTem certeza que deseja remover {pokemon['nome']}? (S/N): ").strip().upper()
        if confirmacao == 'S':
            pokedex.remove(pokemon)
            salvar_pokedex(pokedex)
            print(f"\n{pokemon['nome']} foi removido da sua Pokédex.")
        else:
            print("\nRemoção cancelada.")
    else:
        print(f"\nVocê não tem esse Pokemon.")

def atualizar_nivel(alvo_digitado):
    pokedex = carregar_pokedex()
    pokemon = buscar_pokemon(pokedex, alvo_digitado)

    if pokemon:
        print(f"\nEncontrado: {pokemon['nome']} (Nível atual: {pokemon['nivel']})")
        novo_nivel = obter_nivel_valido("Para qual nível ele subiu? ")

        pokemon['nivel'] = novo_nivel
        salvar_pokedex(pokedex)
        print(f"\n{pokemon['nome']} agora está nível {novo_nivel}.")
    else:
        print(f"\nVocê não tem esse Pokemon.")

def consultar_pokemon_api(alvo_recebido=None):
    if alvo_recebido:
        alvo = alvo_recebido.lower()
    else:
        alvo = input("\nDigite o nome ou ID do Pokémon: ").strip().lower()

    print(f"Consultando a base de dados...")
    url = f"https://pokeapi.co/api/v2/pokemon/{alvo}/"

    try:
        resposta = requests.get(url, timeout=10)

        if resposta.status_code == 200:
            try:
                dados = resposta.json()
            except json.JSONDecodeError:
                print("\nErro: Resposta inválida da API.")
                return

            tipos = [t['type']['name'].capitalize() for t in dados['types']]

            hp = dados['stats'][0]['base_stat']
            ataque = dados['stats'][1]['base_stat']
            defesa = dados['stats'][2]['base_stat']
            velocidade = dados['stats'][5]['base_stat']

            def desenhar_barra(valor):
                tamanho = int((valor / 150) * 15)
                tamanho = min(tamanho, 15)
                return ("█" * tamanho) + ("-" * (15 - tamanho))

            texto_fisico = f"{dados['height']/10:.1f}m | {dados['weight']/10:.1f}kg"

            print("\n╔════════════════════════════════════════════════╗")
            print(f"║ FICHA COMPLETA: {dados['name'].upper():<27}    ║")
            print("╠════════════════════════════════════════════════╣")
            print(f"║ PokeID     : #{str(dados['id']).zfill(3):<33}║")
            print(f"║ Tipo(s)    : {' / '.join(tipos):<33} ║")
            print(f"║ Físico     : {texto_fisico:<33} ║")
            print("╟────────────────────────────────────────────────╢")
            print("║ STATUS BASE DE BATALHA                         ║")
            print(f"║ HP         : {desenhar_barra(hp)} {hp:<3}               ║")
            print(f"║ Ataque     : {desenhar_barra(ataque)} {ataque:<3}               ║")
            print(f"║ Defesa     : {desenhar_barra(defesa)} {defesa:<3}               ║")
            print(f"║ Velocidade : {desenhar_barra(velocidade)} {velocidade:<3}               ║     ")
            print("╚════════════════════════════════════════════════╝\n")

        else:
            print(f"\nErro: Esse Pokemon não existe na PokeLibrary (HTTP {resposta.status_code}).")

    except requests.exceptions.Timeout:
        print("\nErro: A conexão demorou muito. Tente novamente.")
    except requests.exceptions.ConnectionError:
        print("\nErro: Não foi possível conectar à PokeAPI. Verifique sua internet.")
    except requests.exceptions.RequestException as e:
        print(f"\nErro de rede: {e}")

def menu_principal():
    limpar_tela()

    while True:
        print("\n" + "="*30)
        print("      MENU POKEDEX")
        print("="*30)
        print("1. Adicionar novo Pokemon")
        print("2. Listar meus Pokemons")
        print("3. Consultar Pokelibrary")
        print("4. Atualizar Pokemon")
        print("5. Remover Pokemon")
        print("0. Sair do sistema")
        print("="*30)

        opcao = input("Escolha uma opção: ")
        limpar_tela()

        if opcao == "1":
            nome_poke = input("\nDigite o nome do Pokemon: ")
            nivel_poke = obter_nivel_valido("Qual o nível dele? ")
            adicionar_pokemon(nome_poke, nivel_poke)

        elif opcao == "2":
            listar_pokemons()
            alvo_digitado = input("Ou digite o ID/Nome de um Pokemon para ver a Ficha ou aperte ENTER para voltar: ").strip()

            if alvo_digitado != "":
                pokedex = carregar_pokedex()
                pokemon = buscar_pokemon(pokedex, alvo_digitado)

                if pokemon:
                    consultar_pokemon_api(alvo_digitado)
                else:
                    print(f"\nVocê não tem esse Pokemon no seu Pokedex!")

        elif opcao == "3":
            consultar_pokemon_api()

        elif opcao == "4":
            listar_pokemons()
            alvo = input("Digite o ID ou o nome do Pokemon que deseja atualizar: ").strip()
            atualizar_nivel(alvo)

        elif opcao == "5":
            listar_pokemons()
            alvo = input("Digite o ID ou o nome do Pokemon que deseja remover: ").strip()
            remover_pokemon(alvo)

        elif opcao == "0":
            print("\nFechando o Pokedex")
            break

        else:
            print("\nOpção inválida! Digite um número do menu.")

        if opcao != "0":
            input("\nPressione ENTER para voltar ao menu.")
            limpar_tela()

# Executa o sistema de forma protegida (boa prática do Python)
if __name__ == "__main__":
    menu_principal()
```
